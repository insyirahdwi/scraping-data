```python
# tes": Memuat semua library yang ku butuhkan
import urllib.request
from urllib.error import HTTPError, URLError
from bs4 import BeautifulSoup
import pandas as pd
from datetime import datetime

print("Library berhasil dimuat")
```

```python
# proses 2:  URL yang ku pakai
url = "https://himabisdigfebunm.com/"

headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) Chrome/120.0.0.0 Safari/537.36'
}

req = urllib.request.Request(url, headers=headers)
print(f"Target URL diset ke: {url}")
```

```python
# proses 3 : Mengunduh HTML dan membuat objek parser
try:
    html = urllib.request.urlopen(req)
    soup = BeautifulSoup(html.read(), "html.parser")
    print("Berhasil mengambil dan memparsing HTML halaman web!")
except HTTPError as e:
    print("HTTP Error:", e)
except URLError as e:
    print("Server tidak ditemukan:", e)
```

```python
# proses 4: Mengambil field data yang ku butuhkan (seperti contoh pak anto)

# 1. Title
title_tag = soup.find('title')
title = title_tag.get_text(strip=True) if title_tag else "N/A"

# 2. Description (desc)
meta_desc = soup.find('meta', attrs={'name': 'description'}) or soup.find('meta', attrs={'property': 'og:description'})
if meta_desc and meta_desc.get('content'):
    desc = meta_desc['content'].strip()
else:
    first_p = soup.find('p')
    desc = first_p.get_text(strip=True) if first_p else "N/A"

# 3. Content
main_content = soup.find('article') or soup.find('main') or soup.body
if main_content:
    paragraphs = [p.get_text(strip=True) for p in main_content.find_all('p') if p.get_text(strip=True)]
    content = " ".join(paragraphs[:3]) if paragraphs else "N/A"
else:
    content = "N/A"

# 4. Timestamp & Created At
time_tag = soup.find('time')
meta_time = soup.find('meta', attrs={'property': 'article:published_time'})

if time_tag and time_tag.get('datetime'):
    timestamp = time_tag['datetime']
elif meta_time and meta_time.get('content'):
    timestamp = meta_time['content']
else:
    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

created_at = timestamp

# 5. User
meta_author = soup.find('meta', attrs={'name': 'author'}) or soup.find('meta', attrs={'property': 'article:author'})
author_tag = soup.find(class_=lambda c: c and 'author' in c.lower()) if not meta_author else None

if meta_author and meta_author.get('content'):
    user = meta_author['content'].strip()
elif author_tag:
    user = author_tag.get_text(strip=True)
else:
    user = "Admin HIMA BISDIG"

print("Proses ekstraksi data selesai!")
```

```python
# proses 5: Menyusun data ke dalam Pandas DataFrame
scraped_data = [{
    'desc': desc,
    'content': content,
    'title': title,
    'timestamp': timestamp,
    'created_at': created_at,
    'user': user
}]

df = pd.DataFrame(scraped_data)

# Tampilkan tabel data di Colab
display(df)
```

```python
# proses 6: Ekspor data ke file CSV
csv_filename = "data_scraping_himabisdig.csv"
df.to_csv(csv_filename, index=False)

print(f"File berhasil disimpan dengan nama: {csv_filename}")
```
